```python
#Step 0: Import Libraries

# Import Pandas for data analysis
import pandas as pd

# Import NumPy for mathematical and statistical operations
import numpy as np
```

```python
#Step 1: Load Dataset


# Read CSV file and store it in a DataFrame
df = pd.read_csv('/content/22_restaurant_orders.csv')

# Display first 5 records
df.head()
```

```python
# Returns (number of rows, number of columns)
df.shape
```

```python
# Display column names
df.columns
```

```python
# Display column names, data types and non-null values
df.info()
```

```python
# Generate statistical summary of numerical columns
df.describe()
```

```python
#Step 2: Sales Analysis

# Calculate revenue for each order
df['Revenue'] = df['Quantity'] * df['Unit_Price']

# Display first 5 records
df.head()
```

```python
#Calculate total revenue

# Calculate revenue for each order
df['Revenue'] = df['Quantity'] * df['Unit_Price']

# Sum all revenue values
total_revenue = df['Revenue'].sum()

print("Total Revenue:", total_revenue)
```

```python
#Calculate average revenue

# Calculate revenue for each order
df['Revenue'] = df['Quantity'] * df['Unit_Price']

# Calculate average revenue per order record
average_revenue = df['Revenue'].mean()

print("Average Revenue:", average_revenue)
```

```python
#Find maximum revenue

# Calculate revenue for each order
df['Revenue'] = df['Quantity'] * df['Unit_Price']

# Find the highest revenue generated by a single order record
maximum_revenue = df['Revenue'].max()

print("Maximum Revenue:", maximum_revenue)
```

```python
#Find minimum revenue

# Calculate revenue for each order
df['Revenue'] = df['Quantity'] * df['Unit_Price']

# Find the lowest revenue generated by a single order record
minimum_revenue = df['Revenue'].min()

print("Minimum Revenue:", minimum_revenue)
```

```python
#Step 3: Menu Analysis


# Display all unique dish names
df['Dish_Name'].unique()
```

```python
# Count order for each dish
dish_orders = df['Dish_Name'].value_counts()

print(dish_orders)
```

```python
# Find the most popular dish
most_popular_dish = df['Dish_Name'].value_counts().idxmax()

print("Most Popular Dish:", most_popular_dish)
```

```python
# Group by dish and calculate total revenue
dish_revenue = df.groupby('Dish_Name')['Revenue'].sum()

# Sort revenue in descending order
dish_revenue = dish_revenue.sort_values(ascending=False)

print(dish_revenue)
# Identify the dish generating the highest revenue
top_dish = dish_revenue.idxmax()

print("Top Revenue Generating Dish:", top_dish)
```

```python
# Calculate average unit price for each dish
average_dish_price = df.groupby('Dish_Name')['Unit_Price'].mean()

print(average_dish_price)
```

```python
#Step 4: Restaurant Analysis

# Display all restaurant names
df['Restaurant_Name'].unique()
```

```python
# Count order records for each restaurant
restaurant_orders = df['Restaurant_Name'].value_counts()

print(restaurant_orders)
```

```python
#  calculate revenue restuarant wise
restaurant_revenue = df.groupby('Restaurant_Name')['Revenue'].sum()

# Sort revenue from highest to lowest
restaurant_revenue = restaurant_revenue.sort_values(ascending=False)

print(restaurant_revenue)
```

```python
#Step 5: Cuisine Analysis

# Display unique cuisines
df['Cuisine'].unique()
```

```python
# Count order records for each cuisine
cuisine_orders = df['Cuisine'].value_counts()

print(cuisine_orders)
```

```python
#Step 6: Category Analysis


# Display all available categories
df['Category'].unique()
```

```python
# Count orders belonging to each category
category_orders = df['Category'].value_counts()

print(category_orders)
```

```python
# Calculate total quantity sold in each category
category_quantity = df.groupby('Category')['Quantity'].sum()

print(category_quantity)
```
